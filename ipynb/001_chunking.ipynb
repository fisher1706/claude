```python
# Chunk by a set number of characters
def chunk_by_char(text, chunk_size=150, chunk_overlap=20):
    """
    Split text into chunks of specified character count with optional overlap.

    Args:
        text: Input text to chunk
        chunk_size: Number of characters per chunk
        chunk_overlap: Number of overlapping characters between consecutive chunks

    Returns:
        List of text chunks
    """
    # Input validation
    if not text or chunk_size <= 0:
        return []

    # Ensure overlap doesn't exceed chunk_size - 1 to prevent infinite loops
    overlap = min(chunk_overlap, chunk_size - 1) if chunk_overlap > 0 else 0

    chunks = []
    start_idx = 0

    while start_idx < len(text):
        # Calculate end index (don't exceed text length)
        end_idx = min(start_idx + chunk_size, len(text))

        # Extract and store chunk
        chunk_text = text[start_idx:end_idx]
        chunks.append(chunk_text)

        # Break if we've reached the end of text
        if end_idx >= len(text):
            break

        # Move start index: advance by chunk_size minus overlap
        start_idx = end_idx - overlap

    return chunks
```

```python
# Chunk by sentence
import re


def chunk_by_sentence(text, max_sentences_per_chunk=5, overlap_sentences=1):
    sentences = re.split(r"(?<=[.!?])\s+", text)

    chunks = []
    start_idx = 0

    while start_idx < len(sentences):
        end_idx = min(start_idx + max_sentences_per_chunk, len(sentences))

        current_chunk = sentences[start_idx:end_idx]
        chunks.append(" ".join(current_chunk))

        start_idx += max_sentences_per_chunk - overlap_sentences

        if start_idx < 0:
            start_idx = 0

    return chunks
```

```python
# Chunk by section
def chunk_by_section(document_text):
    pattern = r"\n## "
    return re.split(pattern, document_text)
```

```python
with open("report.md", "r") as f:
    text = f.read()

chunks = chunk_by_char(text, chunk_size=150, chunk_overlap=20)

print(f"Total text length: {len(text)} characters")
print(f"Number of chunks: {len(chunks)}")
print(f"Average chunk size: {sum(len(c) for c in chunks) / len(chunks):.1f} characters")
print("\nFirst 3 chunks:")
print("=" * 60)
for i, chunk in enumerate(chunks[:3], 1):
    print(f"\nChunk {i} ({len(chunk)} chars):")
    print(chunk)
    print("-" * 60)
```
